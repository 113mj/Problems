#variable Naming (변수 이름 규칙)

- 변수의 이름을 지을 때의 관례
-지금은 지키지 않아도 상관이 없는데, 나중에 이 규칙을 지키지 않으면 오류가 나는 경우가 있음.

1. snake_case
- 단어와 단어 사이를 언더바(_)를 이용해서 이어줍니다.
- student_name

2. CamelCase
- 단어와 단어를 이어줄 때 뒤에 오는 단어의 첫 글자를 대문자로 이어준다.
- StudentName or studentName
  - 앞 글자부터 대문자: 클래스(class) 이름을 지을 때
  - 뒤 글자부터 대문자: 변수의 이름을 지을 때
  
3. kebab-case
- 단어와 단어 사이를 하이푼(-)을 이용해서 이어줍니다.
- student-name
파이썬의 관례상 변수의 이름은 snake_case, 클래스의 이름은 CamelCase

```python
student_name = "minho"
```

```python
class StduentName:
  pass
```

```python
학생이름 = "민호"
print(학생이름)
```
